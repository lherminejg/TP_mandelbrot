# Ensemble de Mandelbrot


Le but de ce TP est de représenter [l'ensemble de Mandelbrot](https://fr.wikipedia.org/wiki/Ensemble_de_Mandelbrot).

En mathématiques, l'ensemble de Mandelbrot est une fractale définie comme l'ensemble des points $c$ du plan complexe pour lesquels la suite de nombres complexes définie par récurrence par :
$\begin{cases} 
z_0 = 0 \\
z_{n+1} = z_n^2+c
\end{cases}$

est bornée.
\
On démontre que lorsque $|z|>2$, la suite diverge.

Il s'agit ici de compléter la fonction <code>mandelbrot(height, width)</code> pour reproduire la figure ci-dessus.

- Partir d’un pavé rectangulaire des complexes : $\Re(z) \in [-2 ; 0,8]$ et $\Im(z) \in [-1,4 ; 1,4]$.
- Construire un tableau <code>c</code> contenant une découpe de ce pavé en un maillage de $height \times width$ (ici, $1000 \times 1000$).
- Calculer pour chaque cellule du tableau la première itération $n$ où la suite diverge (son module est supérieur à $2$).
On se fixe ici une valeur maximale pour les itérations : $n<n_{max}=20$). On stockera les valeurs de $n$ dans un tableau <code>diverge</code>.
- Afficher l'image obtenue avec <code>plt.imshow(diverge)</code>.

**Rappel:** Ne pas utiliser de boucles for pour parcourir les éléments du tableau, il faut utiliser la vectorisation.

*Indications:* 
- *regarder les fonctions <code>np.linspace</code>, <code>np.meshgrid</code>.*
- *Il est utile de noter que numpy possède un type <code>complex</code>.*
- *Pensez aux masques*

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# à vous de jouer
def mandelbrot(height, width):
    R = np.linspace(-2, 0.8, width) 
    I = np.linspace(-1.4,1.4,height).reshape(height,1)
    tab_c = R + I * 1j #tableau des coordonnées des points c

    tabZ = np.zeros((height,width)) #tableau des suites z de tab_c
    diverge = np.zeros((height,width)) #on stocke ici les n des points dont la suite diverge
    for n in range (20): 
        tabZ = tabZ**2 + tab_c
        mask = (np.abs(tabZ)>2) & (diverge == 0) #on crée le masque qui identifie les points pour lesquels la suite va diverger donc tq |z|>2
        diverge[mask]= n+1 #on rajoute ces points aux points qu'on considère qui divergent
    return diverge
```

```python
mandelbrot(1000,1000)
```

```python
# et pour la tester, pour produire la mème figure que ci-dessus
diverge = mandelbrot(1000, 1000)
plt.imshow(diverge)
plt.show()
```
